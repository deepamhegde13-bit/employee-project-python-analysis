## Task 1 – Create Employee, Project and Seniority DataFrames

### Objective

Create three separate Pandas DataFrames for employees, projects, and seniority levels. Export each DataFrame as a CSV file to be used in subsequent business analysis tasks.

---

```python
# ==================================================
# Task 1 : Create Employee, Project and Seniority DataFrames
# ==================================================

# Import the Pandas library
import pandas as pd

# Creating Employee DataFrame containing employee details.
Employee = pd.DataFrame({"ID": ["A001","A002","A003","A004","A005"], "Name": ["John Alter","Alice Luxumberg","Tom Sabestine","Nina Adgra","Amy Johny"],
                         "Gender": ["M","F","M","F","F"], "City": ["Paris","London","Berlin","Newyork","Madrid"], "Age": [25, 27, 29, 31, 30]})

# Creating Seniority Level DataFrmae with ID and Designation Level.
Seniority = pd.DataFrame({"ID": ["A001","A002","A003","A004","A005"], "Designation Level": [2,2,3,2,3]})

# Creating Project DataFrame with ID, Project, Cost and Status.
Project = pd.DataFrame({"ID": ["A001","A002","A003","A004","A005","A002","A005","A003","A001","A003","A001","A004","A004","A005"], "Project": ["Project 1","Project 2","Project 3","Project 4","Project 5","Project 6","Project 7","Project 8","Project 9","Project 10","Project 11","Project 12","Project 13","Project 14"],
                        "Cost": [1002000,2000000,4500000,5500000,None,680000,400000,350000,None,300000,2000000,1000000,3000000,200000], "Status": ["Finished","Ongoing","Finished","Ongoing","Finished","Failed","Finished","Failed","Ongoing","Finished","Failed","Ongoing","Finished","Finished"]})
```

```python
# Save all three DataFrames to .CSV file
Employee.to_csv("Employee.csv", index=False)
Seniority.to_csv("Seniority.csv", index=False)
Project.to_csv("Project.csv", index=False)
```

## Task 2 – Handle Missing Project Costs

### Objective

Replace missing values in the **Project Cost** column using the running average of previous project costs to ensure data completeness for business analysis.

---
The following code reads the project dataset and replaces missing project costs using a running average approach.

```python
# ==================================================
# Task 2 : Handle Missing Project Costs
# ==================================================

Project = pd.read_csv("Project.csv")

# Initialize total and count for running average
total = 0
count = 0

# Loop through each row in the DataFrame
for i in range(len(Project)):
    # Check if Cost is NOT NaN
    if pd.notna(Project.loc[i, "Cost"]):
        # Add cost to total
        total += Project.loc[i, "Cost"]
        # Increase count
        count += 1
    else:
        # Replace NaN with running average
        Project.loc[i, "Cost"] = total/count

# Save updated DataFrmae
Project.to_csv("UpdatedProject.csv", index=False)

# Display updated DataFrame
Project
```

## Task 3 – Split Employee Names

### Objective

Separate employee names into **First Name** and **Last Name** columns to improve data organization and support reporting.

---
The following code separates employee names into individual first and last name columns for easier analysis.

```python
# ==================================================
# Task 3 : Split Employee Names
# ==================================================

Employee = pd.read_csv("Employee.csv")

# Split the employee name into First Name and Last Name columns.
Employee[["First Name", "Last Name"]] = Employee["Name"].str.split(" ",expand=True)

# Remove the original Name column after splitting.
Employee = Employee.drop("Name", axis=1)

# Save updated DataFrmae
Employee.to_csv("UpdatedEmployee.csv", index=False)

# Display updated DataFrame
Employee
```

## Task 4 – Merge Employee, Seniority and Project Data

### Objective

Combine the employee, seniority, and project datasets into a single DataFrame using the **Employee ID**. Save the merged dataset for use in subsequent business analysis tasks.

---
The following code loads the updated datasets, merges them using the common **ID** column, and exports the consolidated dataset as a CSV file.

```python
# ==================================================
# Task 4 : Merge Employee, Seniority and Project Data
# ==================================================

Employee = pd.read_csv("UpdatedEmployee.csv")
Seniority = pd.read_csv("Seniority.csv")
Project = pd.read_csv("UpdatedProject.csv")

# Merge the Employee and Seniority DataFrames using Employee ID.
Final = pd.merge(Employee, Seniority, on="ID")

# Merge the resulting DataFrame with the Project DataFrame.
Final = pd.merge(Final, Project, on="ID")

# Export the merged DataFrame as a CSV file.
Final.to_csv("Final.csv", index=False)

# Display the final merged DataFrame.
display(Final)
```

## Task 5 – Calculate Employee Bonus

### Objective

Calculate a **5% bonus** for employees whose project status is **Finished**. Employees working on ongoing or failed projects are assigned a bonus of zero.

---
The following code uses NumPy's conditional function to calculate the bonus and updates the final dataset.

```python
# ==================================================
# Task 5 : Calculate Employee Bonus
# ==================================================

import numpy as np

# Calculate a 5% bonus for employees with completed projects.
Final["Bonus"] = np.where(Final["Status"] == "Finished",
                          Final["Cost"]*0.05, 0)

# Save the updated DataFrame as a CSV file.
Final.to_csv("Final.csv", index=False)

# Display the updated DataFrame.
display(Final)
```

## Task 6 – Update Employee Designation Levels

### Objective

Increase the **Designation Level** by one for employees whose project status is **Failed**. Remove any employees whose designation level exceeds the maximum allowed level after the update.

---
The following code updates designation levels based on project status and filters the dataset according to the business rule.

```python
# ==================================================
# Task 6 : Update Employee Designation Levels
# ==================================================

# Increase the designation level by one for employees with failed projects.
Final.loc[Final["Status"] == "Failed", "Designation Level"] +=1

# Remove employees whose designation level exceeds 4.
Final = Final[Final["Designation Level"]<=4]

# Save the updated DataFrame as a CSV file.
Final.to_csv("Final.csv", index=False)

# Display the updated DataFrame.
display(Final)
```

## Task 7 – Add Employee Titles

### Objective

Add appropriate titles (**Mr.** or **Mrs.**) to employee first names based on gender. After adding the titles, remove the **Gender** column since it is no longer required for reporting.

---

```python
# ==================================================
# Task 7 : Add Employee Titles
# ==================================================

# Add titles to employee first names based on gender.
Final.loc[Final["Gender"] == "M", "First Name"] = "Mr. " + Final["First Name"]
Final.loc[Final["Gender"] == "F", "First Name"] = "Mrs. " + Final["First Name"]

# Remove the Gender column after adding titles.
Final = Final.drop("Gender", axis=1)

# Save the updated DataFrame as a CSV file.
Final.to_csv("Final.csv", index=False)

# Display the updated DataFrame.
display(Final)
```

## Task 8 – Promote Employees Based on Age

### Objective

Promote employees who are **older than 29 years** by improving their designation level according to the business rule.

The updated designation levels are then saved to the final dataset.

---

```python
# ==================================================
# Task 8 : Promote Employees Based on Age
# ==================================================

# Iterate through each employee record.
for i in range(len(Final)):
    # Promote employees older than 29 years.
    if Final.loc[i, "Age"]>29:
        # Reduce the designation level by one (promotion).
        Final.loc[i, "Designation Level"] -= 1

# Save the updated DataFrame as a CSV file.
Final.to_csv("Final.csv", index=False)

# Display the updated DataFrame.
display(Final)
```

## Task 9 – Generate Project Cost Summary

### Objective

Calculate the total project cost incurred by each employee by grouping records based on employee ID and name.

This summary provides a consolidated view of project spending per employee.

---

```python
# ==================================================
# Task 9 : Generate Project Cost Summary
# ==================================================

# Calculate the total project cost for each employee.
TotalProjCost = Final.groupby(["ID", "First Name"])["Cost"].sum().reset_index()

# Rename the aggregated Cost column.
TotalProjCost.rename(columns={"Cost":"Total Cost"}, inplace=True)

# Display the summarized project cost.
display(TotalProjCost)
```

## Task 10 – Filter Employees by City

### Objective

Retrieve all employee records where the **City** contains the letter **'o'** (case-insensitive).

This demonstrates text filtering using Pandas string functions.

---

```python
# ==================================================
# Task 10 : Filter Employees by City
# ==================================================

# Filter employees whose city contains the letter 'o'.
filtered_city = Final[Final["City"].str.contains("o", case=False)]

# Display the filtered employee records.
display(filtered_city)
```
